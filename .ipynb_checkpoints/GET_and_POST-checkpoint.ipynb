```python
import requests
```

**GET method**

```python
# no parameters/ NO args
url1 = "http://127.0.0.1:5000"
response = requests.get(url=url1)
print(response.status_code)
print(response.text)
```

**POST METHOD** 

```python
url1 = "http://127.0.0.1:5000"
response = requests.get(url=url1)
response.status_code
```

```python
response.text
```

```python
url2 = "http://127.0.0.1:5000/add1"
response = requests.get(url=url2)
print(response.status_code)
print(response.text)
```

```python
url3 = "http://127.0.0.1:5000/add2/10/20"
response = requests.get(url=url3)
print(response.status_code)
print(response.text)
```

```python
url4 = "http://127.0.0.1:5000/add3/10/20"
response = requests.get(url=url4)
print(response.status_code)
print(response.text)
```

```python
# request.args.get
#  requests.get(url, params = dict)

# request.form.get
# requests.post(url, data = dict)
```

```python
url5 = "http://127.0.0.1:5000/add4"
# requests.args.get which means we are requesting the arguments
d = {'number1':10,'number2':20}
response = requests.get(url=url5,params=d)
print(response.status_code)
print(response.text)
```

```python
url6 = "http://127.0.0.1:5000/add5"
# requests.args.get which means we are requesting the arguments
d = {'number1':50,'number2':60}
response = requests.post(url=url6,data=d)
print(response.status_code)
print(response.text)
```
